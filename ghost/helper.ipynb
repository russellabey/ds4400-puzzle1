# TPR and FPR of `estimate` across lam

`estimate(x, y, lam)` from `ghost_circuit.py` (lasso: mean squared error + lam * L1 penalty, intercept not penalized; devices with a nonzero weight are named).

50 log-spaced lam values from 0.01 to 25, circuit seeds 0 to 99.
For each circuit and lam: **TPR** = (named and on the circuit) / (devices on the circuit), **FPR** = (named but not on the circuit) / (devices not on the circuit). Both are then averaged over the 20 circuits.

```python
import numpy as np
from helper import sweep, plot, table

lams = np.logspace(np.log10(0.01), np.log10(25), 50)
results = sweep(lams)
```

```python
table(results)
```

```python
plot(results, 'Mean TPR and FPR vs. lam (100 circuits)', logx=True)
```

## 51 evenly spaced lam values from 10 to 15

Same procedure (circuit seeds 0 to 19, mean TPR and FPR over the circuits), now on `np.linspace(10, 15, 51)` (steps of 0.1).

```python
lams2 = np.linspace(10, 15, 51)
results2 = sweep(lams2)
table(results2)
```
